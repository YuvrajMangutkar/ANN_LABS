```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x=np.array([
    [2,3],
    [4,5],
    [1,1],
    [2,1]
])
y=np.array([1,1,-1,-1])
```

```python
w=np.zeros(2)
bias=0
lr=0.2
epochs=500
for epoch in range(epochs):
    for i in range(len(x)):
        xi=x[i]
        target=y[i]
        net_output=np.dot(w,xi)+bias
        if net_output>=0:
            prd=1
        else:
            prd=0
        if(prd!=target):
            w=w+(lr*xi*target)
            bias=bias+(lr*target)
print(f"final weight:{w}")
print(f"final bias:{bias}")
```

```python
#create decision region
for i in range(len(x)):
    if(y[i]==1):
        plt.scatter(x[i][0],x[i][1],color="green")
    else:
        plt.scatter(x[i][0],x[i][1],color="red")

#create boundary line
x_values=np.linspace(0,6,100)
y_values=-(w[0]*x_values+bias)/w[1]
plt.plot(x_values,y_values)


plt.xlabel("x1")
plt.ylabel("x2")
plt.title("Perceptron Decision Boundary")
plt.grid()
plt.show()
        
```
