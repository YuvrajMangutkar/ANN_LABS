```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x=np.linspace(-10,10,100)
```

```python
def sigmoid(x):
    return 1/(1+np.exp(-x))

def tanh(x):
    return np.tanh(x)

def relu(x):
    return np.maximum(0,x)

def leaky_relu(x):
    return np.where(x>0,x,0.2*x)

def step(x):
    return np.where(x>=0,1,0)

def softmax(x):
    ex=np.exp(x)
    return ex/np.sum(ex)
    

```

```python
plt.figure(figsize=(12,10))

plt.subplot(2,3,1)
plt.plot(x,sigmoid(x),label="Sigmoid")
plt.title("Sigmoid")
plt.grid()


plt.subplot(2,3,2)
plt.plot(x,tanh(x),label="Tanh")
plt.title("Tanh")
plt.grid()

plt.subplot(2,3,3)
plt.plot(x,relu(x),label="ReLU")
plt.title("ReLU")
plt.grid()


plt.subplot(2,3,4)
plt.plot(x,leaky_relu(x),label="Leaky ReLU")
plt.title("Leaky ReLU")
plt.grid()

plt.subplot(2,3,5)
plt.plot(x,step(x),label="Step")
plt.title("Step")
plt.grid()


plt.subplot(2,3,6)
plt.plot(x,softmax(x),label="softmax")
plt.title("Softmax")
plt.grid


plt.tight_layout()
plt.show()
```
