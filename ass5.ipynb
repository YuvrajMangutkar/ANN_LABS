```python
import numpy as np
```

```python
#two pair vectors
x=np.array([
    [1,-1,1],
    [-1,1,-1]
])

y=np.array([
    [1,1],
    [-1,-1]
])
```

```python
#initialize the weight matrix first
w=np.zeros((x.shape[1],y.shape[1]))
for i in range(len(x)):
    w+=np.outer(x[i],y[i])
print(f"final weight matrix is:{w}")
```

```python
def sign(x):
    return np.where(x>=0,1,-1)
```

```python
#input test
x_test=np.array([1,-1,1])

#forward pass
y_out=sign(np.dot(x_test,w))
print(f"output pattern for the {x_test} is : {y_out}")

recall_ptr=sign(np.dot(y_out,w.T))
print(f"recall input pattern for the {y_out} is : {recall_ptr}")
```

```python


import numpy as np
x=np.array([
    [1,-1,1],
    [-1,1,-1]
])

y=np.array([
    [1,1],
    [-1,-1]
])

w=np.zeros((x.shape[1],y.shape[1]))
for i in range(len(x)):
    w+=np.outer(x[i],y[i])
print(f"final weight matrix is:{w}")


def sign(x):
    return np.where(x>=0,1,-1)

test_pattern=np.array([-1,1,-1])
output_pattern=sign(np.dot(test_pattern,w))
print(f"forward pass {test_pattern} ---> {output_pattern}")

input_pattern=sign(np.dot(output_pattern,w.T))
print(f"backward recall {output_pattern} ---> {input_pattern}")
```
