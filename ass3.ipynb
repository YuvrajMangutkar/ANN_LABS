```python
import numpy as np
```

```python

# Step 1: Convert digit to ASCII binary
def digit_to_binary(digit):
    ascii_val = ord(str(digit))   # ASCII value
    binary = format(ascii_val, '08b')  # 8-bit binary
    return np.array([int(bit) for bit in binary])

# Step 2: Create dataset
X = []
y = []

for i in range(10):
    X.append(digit_to_binary(i))
    if i % 2 == 0:
        y.append(0)   # Even
    else:
        y.append(1)   # Odd

X = np.array(X)
y = np.array(y)
```

```python
# Step 3: Initialize weights and bias
weights = np.zeros(8)
bias = 0
learning_rate = 0.1

# Step 4: Training
for epoch in range(10):
    for i in range(len(X)):
        net_input = np.dot(weights, X[i]) + bias
        
        # Step function
        output = 1 if net_input >= 0 else 0
        
        # Update rule
        error = y[i] - output
        weights = weights + learning_rate * error * X[i]
        bias = bias + learning_rate * error
```

```python
# Step 5: Testing
print("Testing:")
for i in range(10):
    x = digit_to_binary(i)
    net_input = np.dot(weights, x) + bias
    prediction = 1 if net_input > 0 else 0
    
    print(f"Digit: {i}, Prediction: {'Odd' if prediction==1 else 'Even'}")
```
