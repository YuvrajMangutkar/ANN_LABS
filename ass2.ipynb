```python

def call_output(x1,x2):
    w1=1;
    w2=-1;
    thr=1;

    weighted_sum=(x1*w1)+(x2*w2)
    if(weighted_sum>=thr):
        return 1;
    
    else:
        return 0;

for i in range(2):
    for j in range(2):
        result=call_output(i,j);
        print(f"{i} | {j} | {result}\n")

        
```
