```python
import numpy as np
x=np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]
])

y=np.array([
    [0],
    [1],
    [1],
    [0]
])

def sig(x):
    return 1/(1+np.exp(-x))

def dsig(x):
   return x*(1-x)

np.random.seed(42)
w1=np.random.rand(2,2)
b1=np.zeros((1,2))

w2=np.random.rand(2,1)
b2=np.zeros((1,1))

epochs=10000
lr=0.1

for epoch in range(epochs):
    z1=np.dot(x,w1)+b1
    a1=sig(z1)

    z2=np.dot(a1,w2)+b2
    a2=sig(z2)

    error=y-a2

    d_a2=error*dsig(a2)
    d_w2=np.dot(a1.T,d_a2)
    d_b2=np.sum(d_a2,axis=0,keepdims=True)

    d_a1=np.dot(d_a2,w2.T)*dsig(a1)
    d_w1=np.dot(x.T,d_a1)
    d_b1=np.sum(d_a1,axis=0,keepdims=True)

    w2+=lr*d_w2
    w1+=lr*d_w1
    b2+=lr*d_b2
    b1+=lr*d_b1

print(a2)
    
    
```
